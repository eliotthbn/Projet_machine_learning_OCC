```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("MQTTEEB-D_cleaned_data.csv", low_memory=False)
```

## Structure

```python
print("Shape:", df.shape)
print(df.dtypes)
print(df.head())
```

## Duplicates and missing values

```python
print("Duplicates:", df.duplicated().sum())
print(df.isna().sum())
```

## Target distribution

```python
order = ["legitimate", "dos", "slowite", "malformed", "bruteforce", "flood"]
counts = df["label"].value_counts().reindex(order)
print(counts)
print((counts / counts.sum() * 100).round(1))
```

## Figure : Class distribution

```python
plt.figure(figsize=(7, 4))
plt.bar(counts.index, counts.values, color="#4C72B0")
for i, v in enumerate(counts.values):
    plt.text(i, v, f"{v / counts.sum() * 100:.1f}%", ha="center", va="bottom", fontsize=9)
plt.title("Figure 1 - Class distribution")
plt.ylabel("Number of packets")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("fig1_class_distribution.png", dpi=150)
plt.show()
```
